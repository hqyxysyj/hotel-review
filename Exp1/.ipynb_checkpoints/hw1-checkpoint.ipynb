````python
import pandas as pd
import numpy as np
import json
import re
import math
import time
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')
import dashscope
from dashscope import Generation
from concurrent.futures import ThreadPoolExecutor, as_completed
````

````python
DASHSCOPE_API_KEY = "sk-f9c9d25bb85b47aea36b7fc8cf8d3a6c"
file_path = 'file/guangzhou_garden_hotel_comments_2023-2025.csv'
df = pd.read_csv(file_path)
df.head()
````

````python
print("数据形状:", df.shape)
print("\n数据类型:")
print(df.dtypes)
print("\n缺失值统计:")
print(df.isnull().sum())
duplicates = df.duplicated(subset=['_id']).sum()
print(f"重复记录数: {duplicates}")
````

````python
# 填充缺失值
df['review_count'] = df['review_count'].fillna('0条点评')
# 日期转换
def parse_chinese_date(date_str):
    if pd.isna(date_str):
        return pd.NaT
    try:
        # 匹配 "2025年4月5日" 格式
        match = re.match(r'(\d{4})年(\d{1,2})月(\d{1,2})日', str(date_str))
        if match:
            year, month, day = map(int, match.groups())
            return datetime(year, month, day)
        return pd.to_datetime(date_str, errors='coerce')
    except:
        return pd.NaT
df['publish_date_parsed'] = df['publish_date'].apply(parse_chinese_date)
````

````python
# 提取数字
def extract_number(s):
    if isinstance(s, str):
        numbers = re.findall(r'\d+', s)
        return int(numbers[0]) if numbers else 0
    return int(s) if isinstance(s, (int, float)) else 0
df['review_count_num'] = df['review_count'].apply(extract_number)
# 计算评论长度
df['comment_len'] = df['comment'].apply(lambda x: len(str(x)))
# 计算对数（加1平滑）
df['log_comment_len'] = df['comment_len'].apply(lambda x: math.log(x + 1))
df['log_useful_count'] = df['useful_count'].apply(lambda x: math.log(x + 1))
````

````python
# 定义房型映射函数
def map_room_type(room_type):
    if not isinstance(room_type, str):
        return '未知'
    rt = room_type.lower()
    if '大床' in rt:
        return '大床房'
    elif '双床' in rt or '标间' in rt:
        return '双床房'
    elif '套房' in rt:
        return '套房'
    elif '行政' in rt:
        return '行政房'
    else:
        return '其他'
df['fuzzy_room_type'] = df['room_type'].apply(map_room_type)
df.head(3)
````

````python
dashscope.api_key = DASHSCOPE_API_KEY
df['quality_score'] = None
def get_quality_score_single(comment, idx):
    prompt = f"""你是一个酒店评论质量评估专家。请根据以下评论内容，评估其质量（0-10分）。
评分标准：
- 9-10分：信息丰富，具体详细
- 7-8分：有一定信息量
- 5-6分：信息量中等
- 3-4分：信息量较少
- 0-2分：几乎没有实质内容

请只返回一个0-10之间的整数分数。

评论内容：{comment}"""
    
    try:
        response = Generation.call(
            model="qwen-turbo",  # 用turbo更快
            prompt=prompt,
            temperature=0.3,
            max_tokens=10,
            result_format="message"
        )
        if response.status_code == 200:
            score_text = response.output.choices[0].message.content.strip()
            numbers = re.findall(r'\d+', score_text)
            return idx, int(numbers[0]) if numbers else 5
        else:
            return idx, 5
    except:
        return idx, 5

# 分批处理参数
batch_size = 100
max_workers = 10
print("并发质量评分")
total = len(df)
results = {}
for batch_start in range(0, total, batch_size):
    batch_end = min(batch_start + batch_size, total)
    batch_results = {}
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for idx in range(batch_start, batch_end):
            future = executor.submit(get_quality_score_single, df.loc[idx, 'comment'], idx)
            futures.append(future)
        for future in as_completed(futures):
            idx, score = future.result()
            batch_results[idx] = score
    results.update(batch_results)
    time.sleep(0.5)  # 批次间休息0.5秒
for idx, score in results.items():
    df.at[idx, 'quality_score'] = score
print(df[['_id', 'quality_score']].head())
````

````python
# 分类体系
CATEGORY_SYSTEM = """
设施类：房间设施、公共设施、餐饮设施
服务类：前台服务、客房服务、退房/入住效率
位置类：交通便利性、周边配套、景观/朝向
价格类：性价比、价格合理性
体验类：整体满意度、安静程度、卫生状况
"""
df['categories'] = None

def get_categories_single(comment, idx):
    prompt = f"""你是一个酒店评论分析专家。请根据以下评论内容，从给定的分类体系中，选择最匹配的1-3个类别。
请以Python列表的字符串形式返回，例如：['设施类', '服务类']

分类体系：
{CATEGORY_SYSTEM}

评论内容：{comment}
只返回列表，不要有其他内容。"""
    
    try:
        response = Generation.call(
            model="qwen-turbo",
            prompt=prompt,
            temperature=0.2,
            max_tokens=100,
            result_format="message"
        )
        if response.status_code == 200:
            result = response.output.choices[0].message.content.strip()
            result = result.replace('```python', '').replace('```', '').strip()
            try:
                categories = eval(result)
                if isinstance(categories, list):
                    return idx, categories
            except:
                found_cats = []
                for cat in ['设施类', '服务类', '位置类', '价格类', '体验类']:
                    if cat in result:
                        found_cats.append(cat)
                return idx, found_cats
            return idx, []
        else:
            return idx, []
    except:
        return idx, []

print("并发类别划分")
total = len(df)
results_cats = {}
for batch_start in range(0, total, batch_size):
    batch_end = min(batch_start + batch_size, total)
    batch_results = {}
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for idx in range(batch_start, batch_end):
            future = executor.submit(get_categories_single, df.loc[idx, 'comment'], idx)
            futures.append(future)
        for future in as_completed(futures):
            idx, cats = future.result()
            batch_results[idx] = cats
    results_cats.update(batch_results)
    time.sleep(0.5)
for idx, cats in results_cats.items():
    df.at[idx, 'categories'] = cats
df[['_id', 'quality_score', 'categories']].head()
````

````python
df['category_summary'] = None
def get_summary_single(comment, categories, idx):
    if not categories:
        return idx, ""
    
    cats_str = ', '.join(categories)
    prompt = f"""你是一个酒店评论分析专家。以下是一段酒店评论，它主要涉及 {cats_str} 这些方面。
请针对这几个方面，为这条评论生成一段简洁的摘要（不超过80字）。

评论内容：{comment}
只返回摘要，不要有其他内容。"""
    
    try:
        response = Generation.call(
            model="qwen-turbo",
            prompt=prompt,
            temperature=0.3,
            max_tokens=150,
            result_format="message"
        )
        if response.status_code == 200:
            return idx, response.output.choices[0].message.content.strip()
        else:
            return idx, ""
    except:
        return idx, ""

print("并发生成摘要")
total = len(df)
results_summary = {}
for batch_start in range(0, total, batch_size):
    batch_end = min(batch_start + batch_size, total)
    batch_results = {}
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for idx in range(batch_start, batch_end):
            future = executor.submit(get_summary_single, df.loc[idx, 'comment'], df.loc[idx, 'categories'], idx)
            futures.append(future)
        for future in as_completed(futures):
            idx, summary = future.result()
            batch_results[idx] = summary
    results_summary.update(batch_results)
    time.sleep(0.5)
for idx, summary in results_summary.items():
    df.at[idx, 'category_summary'] = summary
df[['_id', 'categories', 'category_summary']].head()
````

````python
# 创建新列
df['reverse_query'] = None
def get_query_single(comment, idx):
    prompt = f"""你是一个搜索引擎优化专家。请根据以下酒店评论内容，生成一个简短的反向查询（Reverse Query）。
反向查询应该是一个用户可能会输入的搜索问题或关键词组合（15字以内）。

例如：评论"房间很大，床很舒服" -> 反向查询："大床房舒适"

评论内容：{comment}
只返回查询语句，不要有其他内容。"""
    
    try:
        response = Generation.call(
            model="qwen-turbo",
            prompt=prompt,
            temperature=0.3,
            max_tokens=50,
            result_format="message"
        )
        if response.status_code == 200:
            return idx, response.output.choices[0].message.content.strip()
        else:
            return idx, ""
    except:
        return idx, ""

print("并发生成反向Query")
total = len(df)
results_query = {}
for batch_start in range(0, total, batch_size):
    batch_end = min(batch_start + batch_size, total)
    batch_results = {}
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = []
        for idx in range(batch_start, batch_end):
            future = executor.submit(get_query_single, df.loc[idx, 'comment'], idx)
            futures.append(future)
        for future in as_completed(futures):
            idx, query = future.result()
            batch_results[idx] = query
    results_query.update(batch_results)
    time.sleep(0.5)

for idx, query in results_query.items():
    df.at[idx, 'reverse_query'] = query
df[['_id', 'reverse_query']].head()
````

````python
output_file = 'guangzhou_garden_hotel_comments_processed.csv'
df.to_csv(output_file, index=False, encoding='utf-8-sig')
# 统计信息
print(f"总评论数: {len(df)}")
print(f"平均质量评分: {df['quality_score'].mean():.2f}")
print(f"\n质量分布:")
print(df['quality_score'].value_counts().sort_index())
print(f"\n类别分布:")
all_categories = []
for cats in df['categories'].dropna():
    if isinstance(cats, list):
        all_categories.extend(cats)
from collections import Counter
cat_counts = Counter(all_categories)
for cat, count in cat_counts.most_common():
    print(f"  {cat}: {count}")
print(f"\n评论平均长度: {df['comment_len'].mean():.0f} 字")
````
